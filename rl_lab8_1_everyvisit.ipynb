```python
import gymnasium as gym
import numpy as np
from collections import defaultdict
env = gym.make("FrozenLake-v1", is_slippery=False)
num_episodes = 5000
gamma = 1.0
V = defaultdict(float)
N = defaultdict(int)
for episode_idx in range(num_episodes):
 # Generate episode using random policy
 episode = []
 state, _ = env.reset() # env.reset() now returns (observation, info)
 done = False
 while not done:
  action = env.action_space.sample()
  next_state, reward, terminated, truncated, _ = env.step(action) # env.step() now returns (observation, reward, terminated, truncated, info)
  done = terminated or truncated # Combine terminated and truncated for 'done'
  episode.append((state, reward))
  state = next_state
 G = 0
 # Backward update
 for t in reversed(range(len(episode))):
  s, r = episode[t]
  G = gamma * G + r
  N[s] += 1
  V[s] += (G - V[s]) / N[s]
# Display estimated values
print("\nEvery-Visit Monte Carlo State Values:")
for s in range(env.observation_space.n):
 print(f"State {s}: V = {V[s]:.2f}")
```
