```python

import numpy as np
import gym


ENV_NAME = "Taxi-v3"
env = gym.make("Taxi-v3" )

unwrapped_env = env.unwrapped


num_states = unwrapped_env.observation_space.n
num_actions = unwrapped_env.action_space.n


print(f"Loaded Environment: {"Taxi-v3" }")
print("Number of states:", num_states)
print("Number of actions:", num_actions)

theta = 1e-10


def policy_evaluation(policy, value_table, unwrapped_env, gamma, theta):
    """
    Evaluate the value function under the current policy.
    Iteratively update state values until convergence.
    """
    while True:
        delta = 0

        for state in range(num_states):
            v = 0
            action = policy[state]

            for prob, next_state, reward, done in unwrapped_env.P[state][action]:

                v += prob * (reward + gamma * value_table[next_state])

            delta = max(delta, abs(value_table[state] - v))
            value_table[state] = v

        if delta < theta:
            break
    return value_table


def policy_improvement(value_table, policy, unwrapped_env, gamma):
    """
    Improve the policy using the current value function.
    """
    policy_stable = True

    for state in range(num_states):
        old_action = policy[state]
        action_values = np.zeros(num_actions)

        for action in range(num_actions):
            for prob, next_state, reward, done in unwrapped_env.P[state][action]:
                action_values[action] += prob * (reward + gamma * value_table[next_state])

        policy[state] = np.argmax(action_values)

        if old_action != policy[state]:
            policy_stable = False
    return policy, policy_stable

def policy_iteration(unwrapped_env, gamma, theta):
    """
    Run policy iteration algorithm.
    Returns optimal policy and value function.
    """

    policy = np.random.choice(num_actions, size=num_states)
    value_table = np.zeros(num_states)
    iteration = 0

    while True:
        iteration += 1

        value_table = policy_evaluation(policy, value_table, unwrapped_env, gamma, theta)


        policy, policy_stable = policy_improvement(value_table, policy, unwrapped_env, gamma)


        if policy_stable:
            print(f"Policy stabilized perfectly after {iteration} iterations.")
            return policy, value_table


optimal_policy, optimal_value = policy_iteration(unwrapped_env, gamma, theta)

print("\nOptimal Policy (best action index at each state array index):")
print(optimal_policy)

print("\nOptimal Value Function (long-term value of each state):")
print(optimal_value)
```
