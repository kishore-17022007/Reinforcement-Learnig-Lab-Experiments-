```python
import gym
import numpy as np

env = gym.make("FrozenLake-v1", is_slippery=False)

print("Number of states:", env.observation_space.n)
print("Number of actions:", env.action_space.n)

V = np.zeros(env.observation_space.n)


gamma = 0.9
theta = 1e-6

while True:
    delta = 0


    for s in range(env.observation_space.n):


        v = V[s]


        action_values = []


        for a in range(env.action_space.n):


            action_value = 0


            for prob, next_state, reward, done in env.P[s][a]:

                action_value += prob * (reward + gamma * V[next_state])


            action_values.append(action_value)


        V[s] = max(action_values)


        delta = max(delta, abs(v - V[s]))


    if delta < theta:
        break


policy = np.zeros(env.observation_space.n, dtype=int)

for s in range(env.observation_space.n):
    action_values = np.zeros(env.action_space.n)
    for a in range(env.action_space.n):
        for prob, next_state, reward, done in env.P[s][a]:
            action_values[a] += prob * (reward + gamma * V[next_state])

    policy[s] = np.argmax(action_values)


print("\nOptimal Value Function:")
print(V)

print("\nOptimal Policy (action per state):")
print(policy.reshape(4, 4))

```

```python
import time
import random


BINS = {
    "blue": "Paper & Cardboard Bin",
    "green": "Glass Bin",
    "yellow": "Plastic Bin"
}

def scan_item_color():
    """Simulates a camera sensor detecting a color."""
    colors = ["blue", "green", "yellow", "unknown"]
    return random.choice(colors)

def sort_waste():
    print("Recycling Robot initialized and active...")
    print("------------------------------------------")


    for item_number in range(1, 6):
        print(f"\nScanning item #{item_number}...")
        time.sleep(1.5)


        color = scan_item_color()
        print(f"Sensor detected color: {color.upper()}")


        if color in BINS:
            target_bin = BINS[color]
            print(f"Action: Moving robotic arm to the [{target_bin}].")
        else:
            print("Action: Unknown material! Routing to General Waste Bin.")

        time.sleep(1)

if __name__ == "__main__":
    sort_waste()
    print("\n Sorting batch complete. Robot powering down.")
```
